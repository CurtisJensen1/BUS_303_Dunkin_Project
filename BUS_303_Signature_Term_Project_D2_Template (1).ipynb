**Task 2.13:** 

- Remove the special values inside each column.

Special characters were identified in the Total Fat (g), Saturated Fat (g), Trans Fat (g), Cholesterol (mg), Sodium (mg), Fiber (g), and Protein (g) columns.

You should have identified the "<5" special value in the Cholesterol (mg) column and the "\xaO" in multiple columns. The "<5" is isolated to the Cholesterol (mg) column, so that special value can be removed by just removing it from that one column. For the "\xaO" special character, you could remove it one column at a time, but there is a more efficient way to do it. You can remove it from all the columns at once. This is probably the best way to do it, but it is not exactly beginner-level, so I am providing the solution for you. ```python

````python
# Task 2.13a: Remove the special values inside the Cholesterol column.

# Remove the <5 value from the Cholesterol column

# ENTER CODE BELOW THIS LINE.
# Hint: dnkn2['Cholesterol (mg)'] = (
#    dnkn2['Cholesterol (mg)']
#    .replace(' ', ' '))

dnkn2['Cholesterol (mg)'] = (
    dnkn2['Cholesterol (mg)']
    .replace('<5', '5')
)
````

````python
# Task 2.13b: Remove the "\xaO" special character from all columns:

# ENTER THE CODE BELOW THIS LINE:

# Hint: for col in dnkn2.select_dtypes(include='object').columns:
#    dnkn2[col] = (
#        dnkn2[col]
#        .str.replace(' ', '', regex=False)
#        .str.strip())

for col in dnkn2.columns:
    dnkn2[col] = (
        dnkn2[col]
        .str.replace('\xa0', '', regex=False)
        .str.strip()
    )
    
````
